```python
# Write a program to remove a given item from the set.

my_set = {10, 20, 30, 40, 50}

item = int(input("Enter an item to remove: "))

my_set.remove(item)

print("Set after removing item:", my_set)
```

```python
# Write a program to create an intersection of sets.

set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

intersection = set1.intersection(set2)

print("Intersection of sets:", intersection)
```

```python
# Write a program to create an union of sets.

set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

union = set1.union(set2)

print("Union of sets:", union)
```

```python
# Write a program to find the maximum and minimum value in a set.

my_set = {10, 20, 5, 40, 30}

maximum = max(my_set)
minimum = min(my_set)

print("Maximum value:", maximum)
print("Minimum value:", minimum)
```
