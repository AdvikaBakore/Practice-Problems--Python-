```python
# Q7. Program to print sum of all the digits of given number.
num = int(input("Enter a number: "))
total = 0
 
while num > 0:
    digit = num % 10        # last digit
    total = total + digit
    num = num // 10         # remove last digit
 
print("Sum of digits =", total)
```

```python
# Q8. Find the factorial of a given number.
num = int(input("Enter a number: "))
factorial = 1
 
for i in range(1, num + 1):
    factorial = factorial * i
 
print("Factorial =", factorial)
 
```

```python
# Q9. Python program to check if given number is prime or not.
num = int(input("Enter a number: "))
 
if num <= 1:
    print("The number is not prime")
else:
    prime = True
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break
 
    if prime:
        print("The number is prime")
    else:
        print("The number is not prime")
 
```

```python
# Q10. Python program to print prime numbers between 10 to 99.
for num in range(10, 100):
    prime = True
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break
    if prime:
        print(num, end=" ")
print()
```

```python
# Q12. Program to reverse a given number.
num = int(input("Enter a number: "))
reverse = 0
 
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
 
print("Reverse =", reverse)
```

```python
# Q13. To determine whether a given year is a leap year or not.
year = int(input("Enter a year: "))
 
if year % 400 == 0:
    print(year, "is a leap year")
elif year % 100 == 0:
    print(year, "is not a leap year")
elif year % 4 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
# Q14. In certain university, grades are assigned to the marks obtained in
#      respective subjects. Write a python program to accept Marks in Math
#      and display respective grade as an output using
#      if...elif...elif..else statements (Using multiple conditions).
#      Hint : Relational operators: <, >, <=, >=, ==, !=
#
#          Marks obtained    Grade assigned
#          90-100            O
#          80-89             A+
#          70-79             A
#          60-69             B
#          50-59             C
#          40-49             P
#          Below 40          F
marks = int(input("Enter marks in Math: "))
 
if marks > 100 or marks < 0:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: O")
elif marks >= 80:
    print("Grade: A+")
elif marks >= 70:
    print("Grade: A")
elif marks >= 60:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 40:
    print("Grade: P")
else:
    print("Grade: F")
```

```python
# Q16. Accept Student Name, Roll Number and Marks of the 3 subjects from
#      the user. Calculate the percentage of the marks and display it.
#      Display the Subject with Highest and lowest marks.
student_name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")
maths = float(input("Enter marks in Mathematics: "))
physics = float(input("Enter marks in Physics: "))
chemistry = float(input("Enter marks in Chemistry: "))
 
percentage = (maths + physics + chemistry) / 3     # each subject out of 100
 
print("\nStudent Name:", student_name)
print("Roll Number:", roll_no)
print("Percentage:", round(percentage, 2), "%")
 
# Highest marks
if maths >= physics and maths >= chemistry:
    highest = "Mathematics"
elif physics >= maths and physics >= chemistry:
    highest = "Physics"
else:
    highest = "Chemistry"
 
# Lowest marks
if maths <= physics and maths <= chemistry:
    lowest = "Mathematics"
elif physics <= maths and physics <= chemistry:
    lowest = "Physics"
else:
    lowest = "Chemistry"
 
print("Subject with Highest Marks:", highest)
print("Subject with Lowest Marks:", lowest)
 
 
```
