```python
# List Q9. Accept 20 values from user and save it in list. Perform
#          following operations on it
#          a) count similar elements of list and print their index value
#          b) count even and odd values of list
#          c) count positive and negative values of list.
numbers = []
for i in range(20):
    num = int(input("Enter value: "))
    numbers.append(num)
print("\nList:", numbers)
 
# a) Count similar elements and print their index values
print("\nSimilar elements and their index values:")
checked = []
for i in range(len(numbers)):
    value = numbers[i]
    if value in checked:
        continue
    checked.append(value)
    if numbers.count(value) > 1:
        print(value, "-> count:", numbers.count(value), ", index:", end=" ")
        for j in range(len(numbers)):
            if numbers[j] == value:
                print(j, end=" ")
        print()
 
# b) Count even and odd values
even_count = 0
odd_count = 0
for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print("\nEven values:", even_count)
print("Odd values:", odd_count)
 
# c) Count positive and negative values
positive_count = 0
negative_count = 0
for num in numbers:
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
print("Positive values:", positive_count)
print("Negative values:", negative_count)
```

```python
# List Q10. Accept 10 values from user and save it in list. Perform
#           following operations on it
#           a) Sort list in ascending order using sorted() function and
#              display sorted list
#           b) sort list in descending order using sort() function
#           c) display length of list
numbers = []
for i in range(10):
    num = int(input("Enter value: "))
    numbers.append(num)
print("\nOriginal List:", numbers)
 
# a) Ascending order using sorted()
ascending = sorted(numbers)
print("Ascending Order:", ascending)
 
# b) Descending order using sort()
numbers.sort(reverse=True)
print("Descending Order:", numbers)
 
# c) Length of list
print("Length of List:", len(numbers))
 
 
```

```python
# List Q11. Accept two lists from user and merge them using + in a single
#           list.
list1 = []
list2 = []
 
print("Enter 5 elements for List 1:")
for i in range(5):
    list1.append(int(input("Enter value: ")))
 
print("\nEnter 5 elements for List 2:")
for i in range(5):
    list2.append(int(input("Enter value: ")))
 
merged_list = list1 + list2
print("\nList 1:", list1)
print("List 2:", list2)
print("Merged List:", merged_list)
```

```python
# List Q12. An acronym is a word formed by taking the first letters of the
#           words in a phrase and making a word from them. For example,
#           RAM is an acronym for "random access memory." Write a program
#           that allows the user to type in a phrase and then outputs the
#           acronym for that phrase. Note: the acronym should be all
#           uppercase, even if the words in the phrase are not capitalized.
phrase = input("Enter a phrase: ")
words = phrase.split()
acronym = ""
 
for word in words:
    acronym = acronym + word[0]
 
print("Acronym:", acronym.upper())
```

```python
# List Q13. Write a program to print the abbreviation of a month, given
#           its number.
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
month_number = int(input("Enter month number (1-12): "))
 
if month_number >= 1 and month_number <= 12:
    print("Month abbreviation:", months[month_number - 1])
else:
    print("Invalid month number")
```

```python
# List Q14. Write a Python program to insert, delete, and display elements
#           of a list.
fruit = ["apple", "banana", "cherry", "watermelon"]
print("Original List:", fruit)
 
# Insert
new_item = input("Enter a fruit to insert: ")
fruit.append(new_item)                 # add at end
fruit.insert(1, "mango")               # add at index 1
print("After Insertion:", fruit)
 
# Delete
fruit.remove("banana")                 # delete by value
removed_item = fruit.pop(0)            # delete by index (returns the item)
print("Removed item:", removed_item)
del fruit[0]                           # delete by index using del
print("After Deletion:", fruit)
 
# Display
print("\nDisplaying elements:")
for item in fruit:
    print(item)
 
```

```python
# List Q15. Python program to merge two lists into a dictionary (one list
#           as keys, another as values).
keys = [101, 102, 103, 104]
values = ["Amit", "Rahul", "Priya", "Sneha"]
 
student_dict = {}
for i in range(len(keys)):
    student_dict[keys[i]] = values[i]
 
print("Dictionary:", student_dict)
```

```python
# List Q16. Python program to remove duplicate elements from a list.
a = [1, 2, 3, 3, 4, 5, 5, 6]
 
unique = []
for item in a:
    if item not in unique:
        unique.append(item)
 
print("Original List:", a)
print("List without duplicates:", unique)
 
# Shortcut using set (order may change)
print("Using set():", list(set(a)))
```

```python
# List Q17. Python program to convert two lists into a dictionary using
#           zip().
keys = [101, 102, 103, 104]
values = ["Amit", "Rahul", "Priya", "Sneha"]
 
student_dict = dict(zip(keys, values))
print("Dictionary:", student_dict)
```

```python
# List Q18. Python program to find common elements in two lists.
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
 
common = []
for element in list1:
    if element in list2:
        common.append(element)
 
print("List 1:", list1)
print("List 2:", list2)
print("Common Elements:", common)
```
